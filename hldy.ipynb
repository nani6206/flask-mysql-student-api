# Flask + MySQL Student API

A small practice notebook for the Flask API I built while learning. The runnable API is in `app.py`; this notebook keeps the main ideas and test examples in one place.

## 1. Install packages

Run this in your terminal, not inside the Flask route.

```python
%pip install flask mysql-connector-python
```

## 2. Routes I practised

- `GET /students` — read all students
- `GET /students/<id>` — read one student
- `POST /students` — add a student
- `PUT /students/<id>` — update name and email
- `PATCH /students/<id>` — update the name only in this simple version
- `DELETE /students/<id>` — delete a student

## 3. Example JSON for POST

In Postman, select Body → raw → JSON and send:

```python
{
  "name": "Nag",
  "email": "nag@gmail.com"
}
```

## 4. Example JSON for PUT

PUT sends both fields in this app:

```python
{
  "name": "Nagarjuna",
  "email": "nagarjuna@gmail.com"
}
```

## 5. Example JSON for PATCH

PATCH changes only the name in this app:

```python
{
  "name": "Nagarjuna Updated"
}
```

## 6. Before running

1. Run `schema.sql` in MySQL Workbench.
2. Install the packages from `requirements.txt`.
3. Put your local MySQL password in `app.py` (keep it out of GitHub).
4. Run `python app.py` and test the routes with Postman.

I used this project to practise routing, reading JSON, running SQL queries, and returning JSON responses.
